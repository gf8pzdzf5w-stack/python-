巢狀迴圈（Nested Loop）

```python
for i in range(1,5):
  for j in range(1,5):
    print("★",end=" ")
  print("☆") #結束，避免無限迴圈
```

```python
for i in range(1,5):
  for j in range(i,5):
    print(j,end=" ")
  print(" ")
```

```python
for i in range(1,6):
  for j in range(i,0,-1):
    print(j,end=" ")
  print(" ")
```

```python
def hight_weight(h, w):
  bmi=w/((h/100)**2)
  return bmi
bmi1=hight_weight(170, 60)
bmi2=hight_weight(165, 55)
bmi3=hight_weight(180, 80)
print(f"第一位{bmi1:.2f}\n第二位{bmi2:.2f}\n第三位{bmi3:.2f}")
```

```python
listA = []
while True:
    item = input("請輸入A倉文具，輸入'/'結束：")
    if item == "/":
        break
    listA.append(item)
listB = []
while True:
    item = input("請輸入B倉文具，輸入'/'結束：")
    if item == "/":
        break
    listB.append(item)
total_list = listA + listB
total_list.append("膠水")
print("二倉合併清單：")
print("/".join(total_list))
```

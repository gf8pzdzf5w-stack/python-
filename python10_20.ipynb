```python
def add(p,q,dis=0.05):
  total=p*q
  tax=total*dis
  total_all=total+tax
  return total,tax,total_all
p=int(input("請輸入單價:"))
q=int(input("請輸入數量:"))
x,y,z=add(p,q)
print(f"小計：{x}\n稅金:{y}\n總金額:{z}")
```

```python
def f(n):
    if n == 1:
        return 1
    if n == 2:
        return 2
    return f(n - 1) + f(n - 2)
result = f(10)
print(f"兔子跳到第 10 階共有 {result} 種不同的跳法。")
```

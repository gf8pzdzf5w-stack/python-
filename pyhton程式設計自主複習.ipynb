```python
''' 單向選擇
if 條件式:  #「：」代表「如果前面的條件成立，請進門執行」
    敘述式
else:
    敘述式.  #只有縮進去的程式碼，才歸屬於那個 if 或 else 的管轄範圍；縮排取消、頂格寫的程式碼，就是不管條件成不成立都會執行的普通指令。
'''
num1 = 10
num2 = 20
num3 = 30
max = num1          # 先把擂台主設為 num1 (10)

if num2 > max:      # 20 > 10 成立！
    max = num2      # 換擂台主：max 變成 20

if num3 > max:      # 30 > 20 成立！
    max = num3      # 換擂台主：max 變成 30

print("最大值是:", max) # 輸出 30
```

```python
# 提示 1：準備一個空的存錢筒，初始值給它多少？   0 1 2 3 4 5 6
total = 0

# 提示 2：要數 1 到 5。記得 range(起點, 終點) 是「含頭不含尾」！
# 如果要算到 5，終點要填幾？（5 + 1 = ？）
for i in range(1, 6):
    # 提示 3：把這一圈的 i 加進存錢筒裡
    total += i

# 提示 4：迴圈跑完後印出來（不要縮排，代表跑完才印）
print("總和是：", total)
```

```python
n = int(input("請輸入一個數字: "))
total = 0 # 依然需要這個乾淨的 0 號存錢筒！
for i in range(1, n+1):
 total += i  # 把每一圈的數字加進 total(讓其成變數，為0互加才能當內容列印出來)
print("總和是：",total)
average = total /n
print("平均是：",average)
```

```python
name = ["木棒", "手槍", "霰彈槍", "狙擊槍", "雷射劍"]
power = (10, 30, 50, 80, 100)
for i in range(len(name)):
  print(f"{i + 1} {name[i]}")
while True:
  search = int(input("請輸入武器編號 (0 結束)："))

  if search == 0:
    break
  elif 1 <= search <= 5:
    idx = search - 1
    print(f"{name[idx]} 的攻擊力是 {power[idx]}")
  else:
    print("找不到武器")
avg_power = sum(power) / len(power)
print(f"平均攻擊力：{avg_power}")
```

```python
price = {"蘋果": 50, "香蕉": 30, "牛奶": 60, "麵包": 40}

print("=== 超市商品 ===")
for item, p in price.items():
  print(f"{item} : {p}元")

total = 0

while True:
  name = input("請輸入商品名稱（輸入 / 結束）：")

  if name == "/":
    break
  elif name in price:
    total += price[name]
  else:
    print("查無此商品")

print(f"總金額：{total} 元")
```
